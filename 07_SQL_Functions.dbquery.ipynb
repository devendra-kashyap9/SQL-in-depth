```sql
-- SQL FUNCTION
/* SQL functions are built-in operations or custom code blocks that accept input parameters, process data, and return a specific output value or result set. They reduce query complexity, format outputs, and accelerate data analysis. */

/* Broadly function are classified as:
    1. Single row func -> NAME.lower() like it gets applied to a single row
    2. Multi row func -> GROUP BY, SUM, COUNT, AVG, etc. like it gets applied to a group of rows. */

-- We also have a category of functions called aggregate functions which are used to perform calculations on a group of rows and return a single value. These functions are often used in conjunction with the GROUP BY clause to aggregate data by a specific grouping column. Some examples of aggregate functions include SUM, COUNT, AVG, MAX, and MIN.

-- Some of the most commonly used aggregate functions in SQL are:
-- COUNT: Returns the number of rows in a table or the number of non-NULL values in a column.
-- SUM: Returns the sum of all values in a numeric column

-- we also have nested function in sql where multiple function are used and the result of one function act as a input for another function.
-- LOWER(LEFT('Maria', 2)) -> results "ma"


-- 1. STRING Functions
-- 1.a. Manipulation Function
-- CONCAT
SELECT 
    first_name,
    country,
    concat(first_name, '-' ,country) AS name_country
FROM workspace.default.customers;

-- UPPER And LOWER
SELECT
    first_name,
    country,
    concat(first_name, '-' ,country) AS name_country,
    lower(first_name) AS low_name,
    upper(first_name) AS UP_NAME
FROM workspace.default.customers;

-- TRIM
-- remove leading and trailing spaces
SELECT
    trim("NoSpaces") AS no_space,
    trim("   LeadingSpace") AS lead_space,
    trim("TrailingSpace     ") AS trail_space;

-- LTRIM and RTRIM
-- remove leading and trailing spaces
SELECT
    ltrim("     LeadingSpace") AS lead_space,
    rtrim("TrailingSpace     ") AS trail_space,
    length(lead_space) - length(trail_space) AS diff;

-- REPLACE
SELECT
    '11/05/2026' AS original,
    replace('11/05/2026', '/', '-') AS replaced;

-- length
SELECT
    length("Hello") AS length,
    length("Hello World") AS length,
    length("Hello World") - length("Hello") AS diff;

-- string extraction
SELECT 
    left("databricks", 5) AS left,   -- first 5 letters
    right("databricks", 5) AS right, -- last 5 chars
    substr("databricks", 7, 5) AS substr, 
	--• The first number (7) is the starting position (SQL indexes start at 1, not 0).
	-- • The second number (5) is the length (how many characters to extract).
    substring("databricks", 7, 5) AS substring;

-- ###################################
--            Numeric Functions
-- ##################################

-- ROUND
SELECT
    3.516 AS original,
    round(3.516) AS rounded,
    round(3.516, 0) AS round_0,
    round(3.516, 1) AS round_1,
    round(3.516, 2) AS round_2,
    round(3.516, 3) AS round_3;

-- ABS
SELECT
    -10 AS original,
    abs(-10) AS abs;

-- CEIL and FLOOR
SELECT
    -10.5 AS original,
    ceil(-10.5) AS ceil,
    floor(-10.5) AS floor;

-- #####################################
--             Date and TIme
-- #####################################

/* Date and time in SQL are used to represent and manipulate dates and times. They are commonly used in applications that require tracking events or scheduling tasks. SQL provides a variety of functions for working with dates and times, including date arithmetic, date formatting.

* formate of date in SQL: YYYY-MM-DD
* formate of time in SQL: HH:MM:SS
* formate of datetime in SQL: YYYY-MM-DD HH:MM:

combination of date and time is called Timestamp: YYYY-MM-DD HH:MM:SS */
-- DATE
SELECT
    current_date() AS today;

-- TIME
SELECT
    current_time() AS now;

-- DATETIME
SELECT
    current_timestamp() AS now;

SELECT
    getdate() as today;

-- Part Extraction
-- day, month, year
SELECT
    current_date() AS today,
    day(today) AS day,
    month(current_date()) AS month,
    year(today) AS year;


-- DATEPART
SELECT
    current_date() AS today,
    date_part('year', today) AS year,
    date_part('month', today) AS month,
    date_part('day', today) AS day,
    date_part('dayofweek', today) AS dayofweek,
    date_part('week', today) AS week,
    date_part('quarter', today) AS quarter;

-- DATEFORMAT
SELECT
    current_date() AS today,
    date_format(today, 'yyyy') AS year,
    date_format(today, 'MMMM') AS month,
    date_format(today, 'dd') AS day,
    date_format(today, 'EEEE') AS dayofweek,
    weekofyear(today) AS week,
    date_part('quarter', today) AS quarter;

-- DATE_TRUNC
-- used to reset the value 
SELECT
    current_timestamp() AS timestamp,
    date_trunc('minute', timestamp) AS Second_reset,
    date_trunc('hour', timestamp) AS Minute_reset,
    date_trunc('day', timestamp) AS Hour_reset,
    date_trunc('month', timestamp) AS dday_reset,
    date_trunc('year', timestamp) AS Month_reset;

-- EOMONTH
SELECT
    current_date() AS today,
    last_day(current_date()) AS end_of_month;

-- calculation
-- DATEADD
SELECT
    current_date() AS today,
    date_add(current_date(), 10) AS ten_days_later,
    date_add(current_date(), -10) AS ten_days_earlier;

SELECT
    current_date() AS today,
    date_add(year, 3, today) AS threeYearsLater,
    cast(date_add(month, 5, today) AS date) fiveMonthsLater,
    cast(date_add(day, 1, today) AS date) Tomorrow,
    date_add(today, 1) AS next_day;

-- DATEDIFF
SELECT
    datediff(current_date(), '2023-01-01') AS days_since_2023_01_01,
    datediff('2023-01-01', current_date()) AS days_until_2023_01_01,
    datediff(year, current_date(), '2027-06-23'),
    datediff(month, current_date(), '2027-06-23'),
    datediff(day, current_date(), '2027-06-23');


-- date formatting
-- cast, convert, format
SELECT
    current_date() AS today,
    cast(current_date() AS string) AS string_date,
    cast(current_date() AS timestamp) AS timestamp_date,
    cast(current_date() AS date) AS date_date,
    cast(current_date() AS timestamp_ntz) AS timestamp_ntz_date,
    cast(current_timestamp() AS time) AS time_date,
    unix_date(current_date()) AS int_date,
    cast(unix_date(current_date()) AS float) AS float_date,
    cast(unix_date(current_date()) AS double) AS double_date,
    cast(unix_date(current_date()) AS decimal) AS decimal_date,
    cast(unix_date(current_date()) AS boolean) AS boolean_date,
    cast(cast(current_date() AS string) AS binary) AS binary_date;

-- cast 
SELECT 
    cast(123 AS string) AS Int_to_String,
    cast('123' AS int) AS String_to_Int,
    cast('2026-11-20' AS date);

-- CONVERT (Databricks SQL has no convert() — use to_date/cast instead)
SELECT
    to_date('2023-01-01', 'yyyy-MM-dd') AS string_to_date,
    cast(123 AS string) AS int_to_string;

-- VALIDATION
-- isdate (no isdate() in Databricks SQL — use try_cast ... IS NOT NULL instead)
SELECT
    try_cast('2023-01-01' AS date) IS NOT NULL AS is_date;


-- ############################################
--             NULL FUNCTIONS
-- ############################################
-- handling NULL VALUE
-- Insert some null values in workspace.default.orders
--  INSERT INTO workspace.default.orders (order_id, order_date, cust_id, sales) 
--  VALUES (1005, NULL, 1, 123), 
--         (1006, '2023-01-01', 4, NULL);

-- ISNULL (SQL Server's isnull(col, default) is nvl(col, default) in Databricks SQL)
SELECT
    cust_id,
    nvl(sales, 0) AS sales
FROM workspace.default.orders;

-- COALESCE
SELECT
    cust_id,
    coalesce(sales, 0) AS sales
FROM workspace.default.orders;

-- handling Nulls
-- add last name columns in customers and intentional keep one blank/null
-- ALTER TABLE workspace.default.customers
-- ADD COLUMN last_name STRING;

-- INSERT INTO workspace.default.customers (cust_id, first_name, last_name)
-- VALUES (7, 'Anna', NULL),
--         (8, 'Ben', 'Doe'),
--         (9, 'Carl', 'Smith'),
--         (10, 'Dana', 'Walker'),
--         (11, 'Evan', 'Jones');

-- NULLIF
-- if first_value == second_value return NULL, ELSE first_value
SELECT
    NULLIF(first_name, 'Anna') AS first_name,
    NULLIF(first_name, 'Ben') AS first_name,
    NULLIF(last_name, '') AS last_name
FROM workspace.default.customers;

-- NULLIF
SELECT
    cust_id,
    first_name,
    last_name,
    concat(first_name, ' ', coalesce(last_name, '')) AS full_name,
    coalesce(score, 0) + 10 AS final_score
FROM workspace.default.customers;


-- Sort the table by score in asc and NULL should appear at last
SELECT
    first_name,
    score
FROM workspace.default.customers
ORDER BY CASE WHEN score IS NULL THEN 1 ELSE 0 END, score;


--confusion check
-- creating VIEW and DATA Policy
WITH test AS(
    SELECT 1 ID, 'A' category UNION
    SELECT 2, NULL UNION
    SELECT 3, '' UNION
    SELECT 4, '   ' UNION
)

-- Run together View and SELECT
SELECT 
    ID,
    len(category) AS char_len,
    category AS policy1,
    trim(category) AS policy2,
    nullif(policy2, '') AS policy3,
    coalesce(policy3, 'unkown') AS policy4
FROM test;

```
