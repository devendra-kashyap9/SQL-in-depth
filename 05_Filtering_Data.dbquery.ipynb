```sql
SHOW TABLES;

SELECT *
FROm workspace.default.customers;

-- Comparison Operator
-- equal operator (=)
SELECT *
FROM workspace.default.customers
WHERE country = "Germany";

-- != / <>
SELECT *
FROM workspace.default.customers
WHERE country <> "Germany";

-- >, >=, <, <=
SELECT *
FROM workspace.default.customers
WHERE score <= 500;

--Logical operator
SELECT *
FROM workspace.default.customers
WHERE country = "Germany" AND score >= 500;

--OR
SELECT *
FROM workspace.default.customers
WHERE country = "Germany" OR score >= 500;

-- NOT
SELECT *
FROM workspace.default.customers
WHERE NOT score < 500;

-- range (BETWEEN)
-- Bounds are INCLUSIVE
SELECT *
FROM workspace.default.customers
WHERE score BETWEEN 500 AND 900;

--membership operator
-- In
SELECT *
FROM workspace.default.customers
WHERE country IN ("Germany", "USA");

-- NOT In
SELECT *
FROM workspace.default.customers
WHERE country NOT IN ("Germany", "USA");

-- Search the pattern
-- LIKE
-- % (0 or more) _ (Exactly one)
SELECT *
FROM workspace.default.customers
WHERE first_name LIKE 'M%';

-- _ (Exactly one)
SELECT *
FROM workspace.default.customers
WHERE first_name LIKE 'Mari_';
```
