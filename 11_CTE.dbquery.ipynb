```sql
-- ######################################################
--              CTE Common Table Expression
-- ######################################################
/* A Common Table Expression (CTE) in SQL is a temporary, named result set defined within the execution scope of a single SELECT, INSERT, UPDATE, DELETE, or MERGE statement.

* Advantages of CTE:
    - 1. CTEs can be used to simplify complex queries by breaking them down into smaller, more manageable parts.
    - 2. CTEs can be used to improve query readability and maintainability by providing a clear and concise way to define temporary result sets.
    - 3. CTEs can be used to improve query performance by reducing the number of times a subquery is executed.
    - 4. CTEs can be used to improve query security by limiting the scope of temporary result sets to the current query.
    - 5. CTEs can be used to improve query portability by allowing the same query to be executed on different databases or data sources.
    - 6. CTEs can be used to improve query reusability by allowing the same temporary result set to be used in multiple queries.

* Difference between CTE and Subqueries:
    - 1. CTEs are defined using the WITH clause, while subqueries are defined using parentheses.
    - 2. CTEs can be used in multiple places within a query, while subqueries can only be used once.
    - 3. CTEs can be used to improve query performance by reducing the number of times a subquery is executed, while subqueries are executed each time they are referenced.
    - 4. CTEs can be used to improve query readability and maintainability by providing a clear and concise way to define temporary result sets, while subqueries can be difficult to read and maintain.
    - 5. CTEs can be used to improve query security by limiting the scope of temporary result sets to the current query, while subqueries can be used to access data from other databases or data sources.
    - 6. CTEs can be used to improve query portability by allowing the same query to be executed on different databases or data sources, while subqueries may require modifications to work on different databases or data sources.
 */

-- Standalone CTE
-- total sales of products
-- we can not use ORDER BY with CTE tho we can use it with main query
WITH CTE_Total_Sales AS
(
    SELECT
        productname,
        sum(sales) as totalSales
    FROM workspace.default.sales
    GROUP BY productname
)
SELECT *
FROM CTE_Total_Sales
WHERE totalsales > 500;

-- Main Query
SELECT
    c.first_name,
    coalesce(c.country, 'N/A')
FROM workspace.default.customers c;

-- last order date
SELECT
    productname,
    max(orderDate) as lastOrderDate
FROM workspace.default.sales
GROUP BY productname;


-- Recursive Query
-- generate a Sequence from 1 to 20
-- In Databricks SQL, a self-referencing CTE requires WITH RECURSIVE
-- (OPTION (MAXRECURSION n) is T-SQL syntax; Databricks uses MAX RECURSION LEVEL n on the CTE itself)
-- Anchor Query only run once
WITH RECURSIVE Series MAX RECURSION LEVEL 25 AS(
    -- Anchor Query
    SELECT 
        1 as myNumber
    UNION ALL
    -- Recursive Quesry
    SELECT
        myNumber + 1
    FROM Series
    WHERE myNumber < 20

)
-- Main Query 
SELECT *
FROM Series;

-- create a table with emp_id, name, salary, department, manager_id
-- CREATE TABLE IF NOT EXISTS workspace.default.details(
--     emp_id INT,
--     name STRING,
--     salary INT,
--     department STRING,
--     manager_id INT
-- );

-- INSERT INTO workspace.default.details (emp_id, name, salary, department, manager_id)
-- VALUES (1, 'John', 50000, 'HR', NULL),
--        (2, 'Jane', 60000, 'IT', 1),
--        (3, 'Bob', 70000, 'Sales', 1),
--        (4, 'Alice', 80000, 'IT', 2),
--        (5, 'Charlie', 90000, 'Sales', 3),
--        (6, 'Dave', 100000, 'HR', 2);

-- Create the employees Hierarchy
-- SELECT *
-- FROM workspace.default.details;
WITH RECURSIVE EmployeeHierearchy AS (
    SELECT
        emp_id,
        name,
        manager_id,
        1 as level
    FROM workspace.default.details
    WHERE manager_id IS NULL
    UNION ALL
    SELECT
        e.emp_id,
        e.name,
        e.manager_id,
        eh.level + 1
    FROM workspace.default.details e
    INNER JOIN EmployeeHierearchy eh 
    ON e.manager_id = eh.emp_id
)
SELECT *
FROM EmployeeHierearchy;

















```
