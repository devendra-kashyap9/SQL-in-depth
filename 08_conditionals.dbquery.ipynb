```sql
/* SQL CASE Statement
The CASE statement in SQL is used to apply conditional logic in SQL queries. It works similarly to IF-ELSE conditions in programming languages.

It evaluates conditions and returns a value based on which condition is satisfied.




Scenario                                                                                              USE Case
------------	                                                                            ---------------------------
Data Transformation:                                                        	Converting short codes (e.g., 'M' / 'F') into full labels ('Male' / 'Female').
Conditional Bucketing:                                                      	Grouping continuous numbers (like salaries or ages) into tiers or brackets.
Dynamic Sorting:                                                            	Customizing ORDER BY so specific priority rows sort to the top.
Safe Calculations:                                                             	Preventing division-by-zero errors by checking if a denominator is 0 before dividing.
 */

SELECT * from workspace.default.orders;

-- case statement
SELECT
    order_id,
    order_date,
    sales,
    CASE
        WHEN sales > 70 THEN 'High'
        WHEN sales > 30 THEN 'Medium'
        ELSE 'Low'
    END AS sales_category
FROM
    workspace.default.orders;

-- aggregation
SELECT
    sales_category,
    sum(sales) AS totalSales
FROM(
    SELECT
    order_id,
    order_date,
    sales,
    CASE
        WHEN sales > 70 THEN 'High'
        WHEN sales > 30 THEN 'Medium'
        ELSE 'Low'
    END AS sales_category
FROM
    workspace.default.orders
)
GROUP BY sales_category
ORDER BY totalSales DESC;

-- abbrevate the country name from customers table
SELECT 
    cust_id,
    first_name,
    CASE
        WHEN first_name LIKE "Mar%" THEN "Sparrow"
        WHEN last_name IS NULL THEN "Potter"
        ELSE last_name
    END AS last_name,
    CASE
        WHEN score > 500 THEN 10
        WHEN score > 300 THEN 5
        ELSE 0
    END AS score_simplified,
    CASE country
        WHEN 'USA' THEN 'US'
        WHEN 'UK' THEN 'UK'
        WHEN 'Germany' THEN 'DE'
        ELSE 'N/A'
    END AS country_abbrevated
FROM workspace.default.customers
ORDER BY cust_id;


-- conditional aggregation
SELECT
    cust_id,
    SUM (CASE
        WHEN sales > 10 THEN 1
        ELSE 0
    END) HighSales,
    count(*) TotalSales
FROM workspace.default.orders
GROUP BY cust_id
ORDER BY cust_id;

--- Aggregate function
-- total number of orders
SELECT 
    count(*)
FROM workspace.default.customers;

-- total sales of all orders
SELECT 
    sum(sales) AS totalSales,
    max(sales) AS HighestSales,
    min(sales) AS LowestSales,
    round(avg( CASE
        WHEN sales IS NULL THEN 0
        ELSE sales
    END), 2) averageSales
FROM workspace.default.orders;

SELECT 
    cust_id,
    count(*) AS nrOfOrder,
    sum(sales) AS totalSales,
    max(sales) AS HighestSales,
    min(sales) AS LowestSales,
    round(avg( CASE
        WHEN sales IS NULL THEN 0
        ELSE sales
    END), 2) averageSales
FROM workspace.default.orders
GROUP By cust_id;


```
