```sql
-- See the available tables
SHOW TABLEs;

-- Delete the table while preserving the table structure
TRUNCATE TABLE workspace.persondetails.persons;

-- DML(Insert, Delete, Update)
-- Inseret
INSERT INTO workspace.persondetails.persons(p_id, person_name, birth_date, phone_number) 
VALUES (1, 'John', '1990-01-01', '1234567890'),
        (2, 'Jane', '1995-05-05', '0987654321'),
        (3, 'Bob', '1980-08-08', '1122334455'),
        (4, 'Alice', '1992-03-03', '5566778899'),
        (5, 'Charlie', '1985-12-12', '9988776655');

-- adding one more column though this doesn't come under DML Commands
ALTER TABLE workspace.persondetails.persons
DROP COLUMN IF EXISTS (score);

ALTER TABLE workspace.persondetails.persons
ADD COLUMN score INTEGER;

-- Updating the Score 
UPDATE workspace.persondetails.persons
SET score = 0;

-- Select is also considered as DML Command
--SELECT count(DISTINCt person_name) 
--FROM workspace.persondetails.persons;

-- SELECT COUNT(*)
-- FROM workspace.persondetails.persons;

--Inserting data from one table to another table
-- SELECT * 
-- FROM workspace.default.customers;


INSERT INTO workspace.persondetails.persons (p_id, person_name, birth_date, phone_number, score)
SELECT
        id,
        first_name,
        NULL,
        "Unkown_number",
        score
FROM workspace.default.customers;

-- seeing the person table
SELECT *
FROM workspace.persondetails.persons;

-- Update the values
UPDATE workspace.persondetails.persons
SET score = 300,
    phone_number = "1234567890"
WHERE p_id = 4;

-- Delete
DELETE FROM workspace.persondetails.persons
WHERE p_id = 5;

-- MERGE
/* The MERGE statement in SQL is used to perform INSERT, UPDATE, and DELETE operations in a single statement based on a matching condition between two tables.

It is also called an UPSERT operation (UPDATE + INSERT).
MERGE is commonly used in:
       * Data warehousing.
       * ETL processes.
       * Incremental data loading.
       * Synchronizing source and target tables.
       * Updating existing records and inserting new records.
*/ 
MERGE INTO workspace.persondetails.persons AS t
USING (SELECT 1 AS id, 'John' AS name, '1990-01-01' AS birth_date, '1234567890' AS phone_number, 300 AS score) AS s
ON t.p_id = s.id
WHEN MATCHED THEN
  UPDATE SET
    person_name = s.name,
    birth_date = s.birth_date,
    phone_number = s.phone_number,
    score = s.score
WHEN NOT MATCHED THEN
  INSERT (p_id, person_name, birth_date, phone_number, score)
  VALUES (s.id, s.name, s.birth_date, s.phone_number, s.score);

-- seeing the table after the operations
SELECT * 
FROM workspace.persondetails.persons;

```
