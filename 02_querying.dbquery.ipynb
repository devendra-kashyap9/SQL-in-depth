```sql
-- This is single line comment.

-- Describing the TABLE and getting the info SHEMA
DESCRIBE workspace.mydatabase.customers;

-- Querying and retreving from Orders table
-- SELECT and FROM
SELECT * 
FROM workspace.mydatabase.orders;

/* this is multiline comment*/

/* We can also perform Individual column details
to see the exact thibgs */
SELECT sales 
FROM workspace.mydatabase.orders;

-- looking details of customers table
SELECT 
    first_name, 
    country
FROM workspace.mydatabase.customers;

-- filtering the data with condition
-- using WHERE clause
SELECT 
    first_name,
    country
from workspace.mydatabase.customers
WHERE country = 'Germany';


-- another query
SELECT * 
FROM workspace.mydatabase.customers
WHERE score != 0;

-- ORDER BY
/* WE can sort the data in ASC and DESC using ORDER BY*/
SELECT *
FROM workspace.mydatabase.orders
ORDER BY sales DESC;

SELECT *
FROM workspace.mydatabase.customers
ORDER BY score ASC;

-- sorting the data based on multiple columns (NESTED SORTING)
SELECT *
FROM workspace.mydatabase.customers
ORDER BY 
    country ASC,
    score DESC;

-- aggregating the data with GROUP BY
-- grouping the similar data
SELECT 
    country,
    SUM(score) as total_score
FROM workspace.mydatabase.customers
GROUP BY country;

-- total score and total number of customers for each country
SELECT
    country as country_name,
    SUM(score) as total_score,
    COUNT(*) as total_customers  -- COUNT(id) will also work
FROM workspace.mydatabase.customers
GROUP BY country
ORDER BY total_customers ASC;

-- aggreagting the data with HAVING clause
/* 
    WHERE filters individual raw rows before data is grouped, and cannot use aggregate functions like SUM or COUNT.
    HAVING filters aggregated groups after data is grouped, and is specifically used to filter on aggregate calculations.   
*/

SELECT
    country,
    SUM(score) AS total_score
FROM workspace.mydatabase.customers
GROUP BY country
HAVING total_score > 800;

-- difference between WHERE and HAVING
-- it is the filtering the condition at which stage

SELECT
    country,
    SUM(score) AS total_score
FROM workspace.mydatabase.customers
WHERE score > 400
GROUP BY country
HAVING total_score > 800;
 
/* 
    How They Work (Order of Execution)
    When SQL runs a query, it follows a strict logical pipeline:
    1. FROM: Pulls the data from the tables.
    2. WHERE: Drops individual rows that do not match the criteria.
    3. GROUP BY: Groups the remaining rows together.
    4. HAVING: Filters out entire groups based on aggregate values.
    5. SELECT: Pulls the final columns.
    6. ORDER BY: Sorts the output
    7. LIMIT: Limit the nuymber of rows
*/

-- find the average score of each country considering only customers where score != 0 and average score > 430
SELECT
    country,
    AVG(score) AS average_score
FROM workspace.mydatabase.customers
WHERE score !=  0
GROUP BY country
HAVING average_score > 430;


-- DISTINCT (Remove Duplicate)
-- return unique list of countries
SELECT
    DISTINCT country
FROM workspace.mydatabase.customers;

-- TOP (LIMIT)
-- it is like limited number of rows we wanna extract
-- Extract top 3 data from customers table
SELECT *
FROM workspace.mydatabase.customers
LIMIT 3;

-- LIMIT with OFFSET
-- TOP is used in MS SQL Server, LIMIT is used in MySQL
/* Offset / Pagination Syntax: LIMIT shines when building pagination (like page 2 of search results) because it natively pairs with OFFSET to skip a set number of rows before it starts counting */

SELECT *
FROM workspace.mydatabase.customers
LIMIT 3  -- limit the number of rows
OFFSET 2;  -- skipa the number of rows provided in offset

-- TOP 3 customers witb highest score 
SELECT 
    first_name,
    score
FROM workspace.mydatabase.customers
ORDER BY score DESC
LIMIT 3;

-- selct two most recent orders
SELECT *
FROM workspace.mydatabase.orders
ORDER BY order_date DESC
LIMIT 2;

-- static value
SELECT 123 AS static_number;
SELECT "Hello!!" AS static_string;
SELECT 123.456 AS static_decimal;
SELECT True AS static_boolean;

-- Miscellaneous (static value and value from tables)
SELECT 
    id,
    first_name,
    "New_Customer" AS customer_type
FROM workspace.mydatabase.customers;

/* If we run the selected highlighted portion it will run that part of query not the entire query*/
SELECT * 
FROM workspace.mydatabase.customers
WHERE country = "Germany";
    
```
